```python
import sqlite3
import pandas as pd

# Create connection
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create table
cursor.execute('''
CREATE TABLE students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER,
    marks INTEGER,
    grade TEXT
)
''')

# Insert sample data
students = [
    ('Kushi', 21, 91, 'A'),
    ('Arjun', 22, 85, 'A'),
    ('Sara', 20, 75, 'B'),
    ('Imran', 23, 66, 'C'),
    ('Ravi', 21, 82, 'B'),
    ('Meena', 20, 89, 'A'),
    ('Rahul', 22, 58, 'D'),
    ('Nikita', 21, 77, 'B')
]

cursor.executemany('INSERT INTO students (name, age, marks, grade) VALUES (?, ?, ?, ?)', students)

# Queries
print("All Students:")
df_all = pd.read_sql_query("SELECT * FROM students", conn)
print(df_all)

print("\nTop Scoring Students (marks > 85):")
df_top = pd.read_sql_query("SELECT name, marks FROM students WHERE marks > 85", conn)
print(df_top)

print("\nAverage Marks by Grade:")
df_avg = pd.read_sql_query("SELECT grade, AVG(marks) as avg_marks FROM students GROUP BY grade", conn)
print(df_avg)

conn.close()
```
